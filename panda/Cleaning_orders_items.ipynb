```python
df['shipping_limit_date'] = pd.to_datetime(df['shipping_limit_date'])
```

```python
df[['price', 'freight_value']].describe()
```

```python
q1 = df['freight_value'].quantile(0.25)
q3 = df['freight_value'].quantile(0.75)

df['freight_category'] = np.select(
    [
        df['freight_value'] <= q1,
        df['freight_value'] <= q3
    ],
    [
        'Low',
        'Medium'
    ],
    default = 'High'
)
```

```python
df['freight_category'].value_counts()
```

```python
q1 = df['price'].quantile(0.25)
q3 = df['price'].quantile(0.75)

df['price_category'] = np.select(
    [
        df['price'] <= q1,
        df['price'] <= q3
    ],
    [
        'Low',
        'Medium'
    ],
    default = 'High'
)
```

```python
df['price_category'].value_counts()
```
