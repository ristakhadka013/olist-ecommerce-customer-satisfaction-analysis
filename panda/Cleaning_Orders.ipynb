```python
import numpy as np
```

```python
df['handling_days'] = np.where(
    df['carrier_before_approval'] == 0,
    (df['order_delivered_carrier_date'] - df['order_approved_at']).dt.days,
    np.nan
)
```

```python
df.info()
```

```python
df['handling_category'] = None

condition = df['carrier_before_approval'] == 0

q1 = df.loc[condition, 'handling_days'].quantile(0.25)
q3 = df.loc[condition, 'handling_days'].quantile(0.75)

df.loc[condition, 'handling_category'] = np.select(
    [
        df.loc[condition, 'handling_days'] <= q1,
        df.loc[condition, 'handling_days'] <= q3
    ],
    [
        'Short',
        'Medium'
    ],
    default='Long'
)
```
